```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

from langchain_openai import OpenAIEmbeddings, ChatOpenAI
from langchain_chroma import Chroma

```

```python
loader = PyPDFLoader("D:\\AnacondaProj\\GenerativeAI\\GenAI_Series\\GenAI_series\\Data\\data_science_syllabus.pdf")
docs = loader.load()
len(docs)
```

```python
# Spliting Data into chunks

splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)

chunks = splitter.split_documents(docs)
len(chunks)
```

```python
embeddings = OpenAIEmbeddings(model="text-embedding-3-small")
```

```python
vectorstore = Chroma.from_documents(chunks, embeddings)
```

```python
query = 'Machine learing data science'

data = vectorstore.similarity_search(query)
data[0].page_content
```

```python
context = ''
for doc in data:
    context += doc.page_content + '\n'

print(context)
```

```python
llm = ChatOpenAI(model="gpt-5")

# res = llm.invoke(f"""Can u provide me answer based on provided context for my questions {context} and question is {query}""")
```

```python
 ## Chain -Context Generate | Prompt | LLM Response | Strparser

def get_context(query):
    data = vectorstore.similarity_search(query)
    context = ''
    for doc in data:
        context += doc.page_content + '\n'
    return{
        "context": context,
        "question": query
    }
```

```python
from langchain_core.prompts import PromptTemplate
prompt = PromptTemplate.from_template("""Can u provide me answer based on provided context for my questions {context} and question is {question}""")


rag_chain = get_context | prompt | llm
rag_chain
```

```python
res = rag_chain.invoke("What is duration of module 3?")
print(res.content)
```
